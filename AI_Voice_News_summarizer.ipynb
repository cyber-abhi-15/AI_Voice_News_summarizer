```python
import requests
import google.generativeai as genai
from IPython.display import Audio
```

```python
from google.colab import userdata
NEWS_API_KEY = userdata.get("YOUR_NEWS_API_KEY")
GEMINI_API_KEY = userdata.get("YOUR_GEMINI_API_KEY")
MURF_API_KEY = userdata.get("YOUR_MURF_API_KEY")
```

```python
genai.configure(api_key=GEMINI_API_KEY)

model = genai.GenerativeModel("gemini-2.5-flash")
```

```python
url = f"https://newsapi.org/v2/everything?q=ai&language=en&sortBy=publishedAt&apiKey={NEWS_API_KEY}"

response = requests.get(url)
news = response.json()

if news["status"] == "ok" and len(news["articles"]) > 0:

    first_article = news["articles"][0]

    article = (
        first_article.get("description")
        or first_article.get("content")
        or first_article.get("title")
    )

    print("Latest News Article:\n")
    print(article)

else:
    print("No news articles found.")
```

```python
prompt = f"""
Summarize the following news article in 50 short sentences.

Article:
{article}
"""

model = genai.GenerativeModel("gemini-2.5-flash")

result = model.generate_content(prompt)

summary = result.text

print("\nAI Summary:\n")
print(summary)
```

```python
murf_url = "https://api.murf.ai/v1/speech/generate"

headers = {
    "api-key": MURF_API_KEY,
    "Content-Type": "application/json"
}

data = {
    "text": summary,
    "voiceId": "en-US-natalie",
    "format": "mp3"
}

response = requests.post(murf_url, headers=headers, json=data)

result = response.json()

audio_url = result.get("audioFile")

print("\nGenerated Voice URL:")
print(audio_url)
```

```python
audio = requests.get(audio_url)

with open("news_voice.mp3", "wb") as f:
    f.write(audio.content)

Audio("news_voice.mp3")
```
